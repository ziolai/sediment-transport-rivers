### Section 1.2: Scalar Wave Equation 

**One-Dimensional Second Order Scalar Wave Equation** Let $x \in \Omega=(0,1)$ denote the spatial domain. Let $t \in [0,T]$ denote the time interval. Let $c$ denote the velocity of wave propagation. Let $\gamma \geq 0$ and $\gamma_3 \geq 0 $ denote the linear friction and cubic friction coefficient, respectively. Let $f(x,t)$ denote the forcing. The partial differential equation for the unknown $u(x,t)$ we intend to solve can be written as 

$$
\frac{\partial^2 \, u}{\partial t^2} 
+ \gamma \frac{\partial \, u}{\partial t} 
+ \gamma_3 \left( \frac{\partial \, u}{\partial t} \right)^3 
= c^2 \frac{\partial^2 \, u}{\partial x^2} + f(x,t) 
$$ 

(observe the plus sign in front of $\gamma$ and square of propagation speed). The non-linear term $\gamma_3 \left( \frac{\partial \, u}{\partial t} \right)^3$ is a non-linear friction term. We assume that the forcing $f(x,t)$ varries in space with $F_0(x)$ and sinusoidally in time with a driving frequency $\omega_d$. We can thus write 

$$
f(x,t) = F_d(x) \, \sin(\omega_d \, t) \, .
$$

where $F_0(x)$ is a suitably choosen amplitude. For simplicity, we supply homogeneous Dirichlet conditions at both sides of $\Omega$, i.e., we impose that 

$$
u(0,t) = 0 \text{ and }  u(1,t) = 0\, . 
$$

We impose non-zero initial conditions for $u(x,t)$ and zero initial conditions for $\dot{u}(x,t)$, i.e., we impose that

$$
u(x,0) = u_0(x) \text{ and } \dot{u}(x,0) = 0 \, . 
$$

**Reference Solution in Linear Case** See notebook [beginner_analytical.ipynb](./beginner_analytical.ipynb). Here we assume that $\gamma \neq 0$ and that $\gamma_3 = 0$. Due to damping, the solution to the homogeneous part of the equation vanishing due to the damping. In the linear case, the particiular solution is varying harmonically as the driving frequency $\omega_d$. In the non-linear case, higher harmonic disappear to the cubic non-linearity.

**Solve numerically using existing packages** We expect that the linear problem including damping in 1D/2D/3D can be solved numerically using [ApproxFun.jl](https://juliaapproximation.github.io/ApproxFun.jl/stable/) by modifying the [heat equation tutorial](https://juliaapproximation.github.io/ApproxFun.jl/stable/generated/PDE/#Heat-equation).

## Section 4: Transient Solution Method 

Transient solution methods allow to generate reference solutions in case of non-linear problems where an analytical methods fail. 

After spatial discretization in 1D, the following system of second order ordinary differential equations is obtained 

$$
\ddot{\mathbf{u}}(t) = \mathbf{F}(\mathbf{u},\dot{\mathbf{u}}, t) \in {\mathbb R}^{N+1}
$$ 

where $\mathbf{F}(\mathbf{u},\dot{\mathbf{u}},t)$ contains both linear and non-linear terms. In the linear case,  $\mathbf{F}$ depends on $\dot{\mathbf{u}}$ due to damping. This systems is supplied with initial conditions for both $\mathbf{u}(t=0)$ and $\dot{\mathbf{u}}(t=0)$. This system of ordinary differential eqautions can be solved using time-integration. We use [SecondOrderProblems](https://docs.sciml.ai/DiffEqDocs/stable/examples/classical_physics/#Second-Order-Linear-ODE) in DifferentialEquations.jl.

## Section 5: Harmonic Balance Method 

### Section 1.5: 1D Linear PDE Problems  

Should we introduce complex-valued amplitudes? 

We assume a single harmonic anzats with spatial amplitudes $A_c(x)$ and $B_s(x)$ and driving frequency $\omega_d$. The harmonic balance anzats $u^{hb}(x,t)$ is thus of the form   

$$
u^{hb}(x,t) = A_c(x) \, \cos(\omega_d \, t) + B_s(x) \, \sin(\omega_d \, t) \, . 
$$

We assume that the initial conditions do not influence the solution $u^{hb}(x,t)$. Although the problem is non-linear, we assume superposition and thus that both the cos and sin amplitude $A(x)$ and $B(x)$ satisfy the homogeneous Dirichlet boundary conditions. The velocity is then given by 

$$
\dot{u}^{hb}(x,t) = - \omega_d \, A_c(x) \, \sin(\omega_d \, t) + \omega_d \, B_s(x) \, \cos(\omega_d \, t) \, .  
$$

and similar for the acceleration $\ddot{u}^{hb}(x,t)$. Substituting the above expressoion for the non-linear term into the wave equation, we obtain the two following coupled (so-called Helmholtz) equations for $A_c(x)$ and $B_s(x)$ 

$$
c^2 \, A_c''(x) + \omega_d^2 \, A_c + \gamma \, \omega_d \, B_s(x) = 0 
$$ 

and 

$$
c^2 \, B_s''(x) + \omega_d^2 \, B_s - \gamma \, \omega_d \, A_c(x) = F_d(x)
$$ 

See, e.g., Mandel-Scheider-2020. 

**Solve symbolically** Linear boundary value problem for two coupled ODE supplied with Dirichlet boundary conditions on both sides that can be solved symbolically. See notebook [beginner_symbolic.ipynb](./beginner_symbolic.ipynb). Having solved for $A_c(x)$ and $B_s(x)$, the solution $u^{hb}(x,t)$ is found. Approach limited to linear equations in 1D.
